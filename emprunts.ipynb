```python
import mlxtend
import numpy as np
import datetime
import pandas as pd
import matplotlib.pyplot as plt
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
```

```python
data = pd.read_excel('dataset_apriori_bibliotheque_2025_2026.xlsx')
```

```python
ts = datetime.datetime.now()
start = datetime.datetime(ts.year-1,9,1)
end = datetime.datetime(ts.year,1,31)
df1= data[(start<=data["borrowed_at"])&(data["borrowed_at"]<=end)]
df2= data[data["borrowed_at"]>end]
```

```python
df1=df1.groupby(["student_id"]).agg(",".join).reset_index()
```

```python
df2=df2.groupby(["student_id"]).agg(",".join).reset_index()
```

```python
df1=df1.drop(["student_id"],axis=1)
```

```python
df2=df2.drop(["student_id"],axis=1)
```

```python
new= df1["notice_id"].str.split(",", n=4, expand=False)
array_1= new.to_numpy()
```

```python
new_2= df2["notice_id"].str.split(",", n=4, expand=False)
array_2= new.to_numpy()
```

```python
te= TransactionEncoder().set_output(transform="pandas")
te_arr= te.fit(array_1).transform(array_1)
```

```python
te= TransactionEncoder().set_output(transform="pandas")
te_arr_2= te.fit(array_2).transform(array_2)
```

```python
frequent_items = apriori(te_arr,min_support = 0.05,use_colnames=True)
```

```python
frequent_items_2 = apriori(te_arr_2,min_support = 0.05,use_colnames=True)
```

```python
rules = association_rules(frequent_items, metric='confidence',min_threshold=0.01)
```

```python
rules_2 = association_rules(frequent_items_2, metric='confidence',min_threshold=0.01)
```
